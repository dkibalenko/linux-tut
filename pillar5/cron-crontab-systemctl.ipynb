**COMMAND:** `cron` / `crontab`

**PURPOSE (support):** You already know Celery Beat from Django work — **cron is the OS-level ancestor** of that idea:
> **run a task on a schedule**, no code needed to trigger it, no long-running process to babysit.

In support, cron tickets are extremely common precisely _because_ cron fails silently — no error dialog, no stack trace in your terminal, just... nothing happened. **Learning to hunt down _why_ is a core skill**.

**SYNTAX:** `crontab [flags]` for per-user jobs; system-wide jobs live directly as files.

**FLAGS:**

|Flag|Meaning|
|---|---|
|`crontab -l`|list current user's cron jobs|
|`crontab -e`|edit current user's cron jobs (opens `$EDITOR`)|
|`crontab -r`|remove **all** of current user's cron jobs (no confirmation!)|
|`sudo crontab -u USER -l`|list another user's crontab|
|`sudo crontab -u USER -e`|edit another user's crontab|

**CRON SYNTAX — the 5-field schedule:**

```shell
* * * * * command-to-run
│ │ │ │ │
│ │ │ │ └── day of week (0-6, Sunday=0)
│ │ │ └──── month (1-12)
│ │ └────── day of month (1-31)
│ └──────── hour (0-23)
└────────── minute (0-59)
```

**Developer bridge:** Celery Beat's `crontab(minute=0, hour='*/2')` config _is literally modeled on this exact syntax_ — if you've written Celery schedules, you already half-know this.
> **The difference**: Celery Beat needs your Python process running; cron is baked into the OS and fires even if nothing else is running.

**WHY CRON JOBS FAIL SILENTLY (the real support skill):**

|Cause|Why it's silent|How to catch it|
|---|---|---|
|Wrong `PATH`|cron runs with a minimal environment, not your interactive shell's `PATH` — a script that works fine when _you_ run it can fail under cron because `python3`/`node`/etc. "isn't found"|Always use absolute paths in cron commands (`/usr/bin/python3` not `python3`)|
|No output redirection|cron emails output to the user by default, but most systems (including WSL) have no mail setup, so errors vanish into nowhere|Always redirect: `>> /path/log.txt 2>&1`|
|Working directory assumption|cron's working directory is your home dir, not wherever the script "expects" to run from|Use absolute paths in the script, or `cd /full/path &&` before the command|
|Bad cron syntax|A typo in the schedule can silently mean "never runs" or "runs at the wrong time" rather than erroring|`crontab -l` and re-read the fields carefully; test with `* * * * *` first to confirm the job itself works, then fix timing|

**TICKET SCENARIO:** "A nightly backup cron job hasn't produced a file in 3 days." Diagnosis path:

1. `systemctl status cron` — confirm cron itself is running (if it's down, that alone explains everything, back to the `systemctl`/`journalctl` lessons).
2. `crontab -l` — read the exact line, check the schedule syntax and confirm the command uses absolute paths.
3. `journalctl -u cron --since "3 days ago" | grep -i backup` — did cron _attempt_ to run it at all, or did it never fire?
4. If it fired but the backup still didn't happen — the problem isn't cron, it's the script/command itself. Re-run it manually with the exact same command line cron would use, and check for `PATH` or working-directory-dependent failures that only show up under cron's minimal environment.

**System-wide cron locations** (beyond per-user `crontab`):

- `/etc/cron.d/` — individual files, each with the same 5-field-plus-user syntax, good for package-installed jobs
- `/etc/cron.daily/`, `/etc/cron.hourly/`, `/etc/cron.weekly/` — just scripts dropped in, no schedule needed, run by `run-parts` (you actually saw this exact mechanism in your `journalctl` output earlier: `cd / && run-parts --report /etc/cron.hourly`)

**GOTCHA:** `crontab -r` deletes your **entire** crontab instantly, no confirmation prompt — unlike deleting one line via `-e`. Get in the habit of using `-e` even for cleanup, so you only remove what you intend to.
