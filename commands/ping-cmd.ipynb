## Linux `ping` Command
Test the connectivity between a **local host** and **remote hosts**. Troubleshoot network issues.

### Understanding the Basics of `ping` Command
The `ping` command is a **network diagnostic tool** used to test the connectivity between a **local host** and a **remote host**. It sends Internet Control Message Protocol (ICMP) **echo request packets** to the target host and waits for the ICMP echo reply.

```python
dennie@X390:~/linux-tut$ ping 127.0.0.1
PING 127.0.0.1 (127.0.0.1) 56(84) bytes of data.
64 bytes from 127.0.0.1: icmp_seq=1 ttl=64 time=0.079 ms
64 bytes from 127.0.0.1: icmp_seq=2 ttl=64 time=0.043 ms
64 bytes from 127.0.0.1: icmp_seq=3 ttl=64 time=0.032 ms
64 bytes from 127.0.0.1: icmp_seq=4 ttl=64 time=0.042 ms
64 bytes from 127.0.0.1: icmp_seq=5 ttl=64 time=0.022 ms
64 bytes from 127.0.0.1: icmp_seq=6 ttl=64 time=0.031 ms
64 bytes from 127.0.0.1: icmp_seq=7 ttl=64 time=0.036 ms
64 bytes from 127.0.0.1: icmp_seq=8 ttl=64 time=0.035 ms
64 bytes from 127.0.0.1: icmp_seq=9 ttl=64 time=0.021 ms
64 bytes from 127.0.0.1: icmp_seq=10 ttl=64 time=0.078 ms
64 bytes from 127.0.0.1: icmp_seq=11 ttl=64 time=0.073 ms
64 bytes from 127.0.0.1: icmp_seq=12 ttl=64 time=0.080 ms
64 bytes from 127.0.0.1: icmp_seq=13 ttl=64 time=0.153 ms
^C
--- 127.0.0.1 ping statistics ---
13 packets transmitted, 13 received, 0% packet loss, time 12343ms
rtt min/avg/max/mdev = 0.021/0.055/0.153/0.035 ms
```

The `ping` command sends ICMP echo request packets to the loopback address `127.0.0.1` and displays the response time and other statistics. Press `Ctrl+C` to stop the ping.

Now, let's try pinging a remote host, such as `google.com`:

```python
dennie@X390:~/linux-tut$ ping -c 4 google.com
PING google.com (142.251.209.14) 56(84) bytes of data.
64 bytes from mil04s50-in-f14.1e100.net (142.251.209.14): icmp_seq=1 ttl=112 time=27.5 ms
64 bytes from mil04s50-in-f14.1e100.net (142.251.209.14): icmp_seq=2 ttl=112 time=26.8 ms
64 bytes from mil04s50-in-f14.1e100.net (142.251.209.14): icmp_seq=3 ttl=112 time=26.8 ms
64 bytes from mil04s50-in-f14.1e100.net (142.251.209.14): icmp_seq=4 ttl=112 time=27.0 ms

--- google.com ping statistics ---
4 packets transmitted, 4 received, 0% packet loss, time 3031ms
rtt min/avg/max/mdev = 26.781/27.022/27.527/0.302 ms
```

The `ping` command resolves the hostname `google.com` to its IP address `142.250.184.78` and sends ICMP echo request packets to that address. The response time and other statistics are displayed.

### Troubleshooting Network Issues with ping
First, let's try pinging a host that is not reachable:

```python
dennie@X390:~/linux-tut$ ping -c 4 8.8.8.0
PING 8.8.8.0 (8.8.8.0) 56(84) bytes of data.

--- 8.8.8.0 ping statistics ---
4 packets transmitted, 0 received, 100% packet loss, time 3000ms
```

In this case, the `ping` command shows 100% packet loss, indicating that the host is not reachable.

Next, let's try pinging a host with a long response time:

```python
dennie@X390:~/linux-tut$ ping -c 4 slowweb.com
PING slowweb.com (13.248.169.48) 56(84) bytes of data.
64 bytes from a904c694c05102f30.awsglobalaccelerator.com (13.248.169.48): icmp_seq=1 ttl=246 time=89.8 ms
64 bytes from a904c694c05102f30.awsglobalaccelerator.com (13.248.169.48): icmp_seq=2 ttl=246 time=78.5 ms
64 bytes from a904c694c05102f30.awsglobalaccelerator.com (13.248.169.48): icmp_seq=3 ttl=246 time=72.4 ms
64 bytes from a904c694c05102f30.awsglobalaccelerator.com (13.248.169.48): icmp_seq=4 ttl=246 time=77.1 ms

--- slowweb.com ping statistics ---
4 packets transmitted, 4 received, 0% packet loss, time 2999ms
rtt min/avg/max/mdev = 72.371/79.451/89.757/6.374 ms
```

In this case, the `ping` command shows a very high response time, which could indicate a network performance issue.

Finally, let's try pinging a host with a high packet loss:

```python
ping lossy.com
PING lossy.com (93.184.216.34) 56(84) bytes of data.
64 bytes from 93.184.216.34: icmp_seq=1 ttl=54 time=100 ms
64 bytes from 93.184.216.34: icmp_seq=2 ttl=54 time=100 ms
64 bytes from 93.184.216.34: icmp_seq=3 ttl=54 time=100 ms
64 bytes from 93.184.216.34: icmp_seq=4 ttl=54 time=100 ms
^C
--- lossy.com ping statistics ---
4 packets transmitted, 4 received, 50% packet loss, time 3003ms
rtt min/avg/max/mdev = 100.000/100.000/100.000/0.000 ms
```

In this case, the `ping` command shows a 50% packet loss, which could indicate a network issue.
