```python
dennie@HP:~/linux-tut$ ls -l
total 200
-rw-r--r-- 1 dennie dennie  16374 Aug 22 18:05 1part-linux-filesystem.ipynb
-rw-r--r-- 1 dennie dennie   1229 Aug 22 18:08 2part-fhs.ipynb
-rw-r--r-- 1 dennie dennie  17085 Aug 21 21:57 CLAUDE.md
-rw-r--r-- 1 dennie dennie      0 Aug 21 22:06 PROGRESS.md
-rw-r--r-- 1 dennie dennie   1782 Aug 23 09:08 commands.md
-rw-r--r-- 1 dennie dennie 139370 Aug 22 18:17 linux-cheatsheet.pdf
-rw-r--r-- 1 dennie dennie  15918 Aug 22
```

```python
labex:project/ $ ls -l logs
total 92
-rwxr-xr-x 1 labex labex 25526 Aug 23 17:02 access.log
-rwxr-xr-x 1 labex labex 10322 Aug 23 17:02 error.log
-rwxr-xr-x 1 labex labex 51586 Aug 23 17:02 server.log
```

- `ls` stands for "list". It shows you what's inside a directory.
- The `-l` option (that's a lowercase L) tells `ls` to give you **more details about each file**, like its size and when it was last modified.
- `ls -l logs` lists the contents of the logs directory in a **detailed format**.
