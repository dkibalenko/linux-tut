## Linus `ip` Command
[Link](https://labex.io/tutorials/linux-linux-ip-command-with-practical-examples-422741)

The `ip` command is a powerful tool for managing **network interfaces** and **configurations**.
```shell
ip [OPTIONS] OBJECT [COMMAND [ARGUMENTS]]
```
Here, `OBJECT` represents the network entity you want to manage, such as `link` for network interfaces, `addr` for IP addresses, `route` for routing tables, and so on. `COMMAND` is the specific action you want to perform on the object, and `ARGUMENTS` are any additional parameters required by the command.

Some common `ip` command options include:

- `-c`: Enables color output
- `-f`: Specifies the address family (e.g., `inet`, `inet6`, `link`)
- `-h`: Displays the help menu
- `-s`: Provides detailed, verbose output
- `-o`: Outputs the information in a single-line format

Display information about **all network interfaces**:

````python
dennie@X390:~/linux-tut$ sudo ip link show
[sudo] password for dennie:
1: lo: <LOOPBACK,UP,LOWER_UP> mtu 65536 qdisc noqueue state UNKNOWN mode DEFAULT group default qlen 1000
    link/loopback 00:00:00:00:00:00 brd 00:00:00:00:00:00
2: eth0: <BROADCAST,MULTICAST,UP,LOWER_UP> mtu 1500 qdisc mq state UP mode DEFAULT group default qlen 1000
    link/ether 00:15:5d:77:66:a9 brd ff:ff:ff:ff:ff:ff
````

This command shows the status and configuration of **all network interfaces on the system**.

Display IP address information for a **specific interface**:

````python
dennie@X390:~/linux-tut$ sudo ip a show eth0
2: eth0: <BROADCAST,MULTICAST,UP,LOWER_UP> mtu 1500 qdisc mq state UP group default qlen 1000
    link/ether 00:15:5d:77:66:a9 brd ff:ff:ff:ff:ff:ff
    inet 172.30.240.133/20 brd 172.30.255.255 scope global eth0
       valid_lft forever preferred_lft forever
    inet6 fe80::215:5dff:fe77:66a9/64 scope link
       valid_lft forever preferred_lft forever
````

Displays the addresses and status of the `eth0` network interface. Here, `ip a` is shorthand for `ip address`.

- **`2`** — The interface’s index number on this system.
- **`eth0`** — The interface name, often used for a wired or virtual Ethernet connection.
- **`BROADCAST`** — The interface supports sending packets to all devices on the local network.
- **`MULTICAST`** — The interface supports sending packets to a selected group of devices.
- **`UP`** — The interface is enabled.
- **`LOWER_UP`** — The underlying network link is active.
- **`mtu 1500`** — The largest packet size, in bytes, that can be sent without fragmentation.
- **`qdisc mq`** — The outgoing traffic queueing discipline. `mq` means the interface uses multiple transmission queues.
- **`state UP`** — The interface is operational.
- **`group default`** — The interface belongs to the default interface group.
- **`qlen 1000`** — The outgoing packet queue length is set to 1,000.

```shell
link/ether 00:15:5d:77:66:a9 brd ff:ff:ff:ff:ff:ff
```
- **`link/ether`** — The interface uses Ethernet link-layer addressing.
- **`00:15:5d:77:66:a9`** — The interface’s MAC address, used to identify it on the local network segment. This may be a virtual MAC address.
- **`brd ff:ff:ff:ff:ff:ff`** — The Ethernet broadcast MAC address. A frame sent to this address is intended for all devices on the local Ethernet network.

```shell
inet 172.30.240.133/20 brd 172.30.255.255 scope global eth0
```
- **`inet`** — This is an IPv4 address.
- **`172.30.240.133`** — The interface’s IPv4 address.
- **`/20`** — The network prefix length. It corresponds to the subnet mask `255.255.240.0`. In this case, the subnet is `172.30.240.0/20`, with usable host addresses typically ranging from `172.30.240.1` to `172.30.255.254`.
- **`brd 172.30.255.255`** — The IPv4 broadcast address for that subnet.
- **`scope global`** — The address can be used beyond the local interface, within its configured network. This does **not** mean it is necessarily reachable from the public internet.
- **`eth0`** — The interface this address belongs to.

```shell
valid_lft forever preferred_lft forever
```
- **`valid_lft forever`** — The address has no scheduled expiration.
- **`preferred_lft forever`** — The address remains preferred for new connections; it is not scheduled to become deprecated.

```shell
inet6 fe80::215:5dff:fe77:66a9/64 scope link
```
- **`inet6`** — This is an IPv6 address.
- **`fe80::215:5dff:fe77:66a9`** — A link-local IPv6 address, used for communication on the same local network link. Routers generally do not forward link-local traffic.
- **`/64`** — The IPv6 prefix length.
- **`scope link`** — The address is limited to the local link.

The exact output depends on the system. In this example, `eth0` is active and has both an IPv4 address and an IPv6 link-local address.

### Manage Network Interfaces with the `ip` Command

Create a new network interface using the `ip link add` command:

````python
dennie@X390:~/linux-tut$ sudo ip link add name myint type dummy
````

This creates a new dummy interface named `myint`. You can verify the interface was created using the `ip link show` command:

````python
dennie@X390:~/linux-tut$ sudo ip link show
1: lo: <LOOPBACK,UP,LOWER_UP> mtu 65536 qdisc noqueue state UNKNOWN mode DEFAULT group default qlen 1000
    link/loopback 00:00:00:00:00:00 brd 00:00:00:00:00:00
2: eth0: <BROADCAST,MULTICAST,UP,LOWER_UP> mtu 1500 qdisc mq state UP mode DEFAULT group default qlen 1000
    link/ether 00:15:5d:77:66:a9 brd ff:ff:ff:ff:ff:ff
5: myint: <BROADCAST,NOARP> mtu 1500 qdisc noop state DOWN mode DEFAULT group default qlen 1000
    link/ether fe:3b:55:a5:54:e7 brd ff:ff:ff:ff:ff:ff
````

Bring the `myint` interface up using the `ip link set` command:

````python
dennie@X390:~/linux-tut$ sudo ip link set dev myint up
````

You can verify the interface is now in the `UP` state:

````python
dennie@X390:~/linux-tut$ sudo ip link show myint
5: myint: <BROADCAST,NOARP,UP,LOWER_UP> mtu 1500 qdisc noqueue state UNKNOWN mode DEFAULT group default qlen 1000
    link/ether fe:3b:55:a5:54:e7 brd ff:ff:ff:ff:ff:ff
````

Delete the `myint` interface using the `ip link delete` command:

````python
dennie@X390:~/linux-tut$ sudo ip link delete dev myint
````

Verify the interface has been removed:

````python
dennie@X390:~/linux-tut$ sudo ip link show
1: lo: <LOOPBACK,UP,LOWER_UP> mtu 65536 qdisc noqueue state UNKNOWN mode DEFAULT group default qlen 1000
    link/loopback 00:00:00:00:00:00 brd 00:00:00:00:00:00
2: eth0: <BROADCAST,MULTICAST,UP,LOWER_UP> mtu 1500 qdisc mq state UP mode DEFAULT group default qlen 1000
    link/ether 00:15:5d:77:66:a9 brd ff:ff:ff:ff:ff:ff
````

The `myint` interface is no longer present in the output.

### Troubleshoot Network Issues Using the `ip` Command
Checking the routing table using the `ip route` command:

````python
dennie@X390:~/linux-tut$ sudo ip route show
default via 172.30.240.1 dev eth0 proto kernel
172.30.240.0/20 dev eth0 proto kernel scope link src 172.30.240.133
````

This shows the default route and the route for the local network interface.

Next, let's simulate a network issue by disabling the `eth0` interface:

````python
sudo ip link set dev eth0 down
````

Now, try to ping a remote host:

````python
ping 8.8.8.8
````

You should see that the ping fails, as the network interface is down.

To troubleshoot further, you can use the `ip addr` command to check the IP address configuration:

````python
sudo ip addr show eth0

Example output:
2: eth0: <BROADCAST,MULTICAST> mtu 1500 qdisc fq_codel state DOWN group default qlen 1000
    link/ether 02:42:ac:11:00:02 brd ff:ff:ff:ff:ff:ff
    inet 172.17.0.2/16 brd 172.17.255.255 scope global eth0
       valid_lft forever preferred_lft forever
    inet6 fe80::42:acff:fe11:2/64 scope link
       valid_lft forever preferred_lft forever
````

The output shows that the `eth0` interface is in the `DOWN` state.

To bring the interface back up, use the `ip link set` command:

````python
sudo ip link set dev eth0 up
````

Verify the interface is now in the `UP` state:

````python
sudo ip addr show eth0

Example output:
2: eth0: <BROADCAST,MULTICAST,UP,LOWER_UP> mtu 1500 qdisc fq_codel state UP group default qlen 1000
    link/ether 02:42:ac:11:00:02 brd ff:ff:ff:ff:ff:ff
    inet 172.17.0.2/16 brd 172.17.255.255 scope global eth0
       valid_lft forever preferred_lft forever
    inet6 fe80::42:acff:fe11:2/64 scope link
       valid_lft forever preferred_lft forever
````

Now, try pinging the remote host again, and the ping should succeed.
