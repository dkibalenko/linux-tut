## Linux `ps` Command
`ps` (process status) command to monitor and manage running processes on your system. The `ps` command is a fundamental tool for system administrators and developers, providing detailed information about the processes currently running, including their process ID, user, CPU and memory usage, and more.

The `ps` command provides a snapshot of the current processes running on the system, including their process ID (PID), user, CPU and memory usage, and other relevant information.

Let's start by running the basic `ps` command:

```python
dennie@HP:~/linux-tut$ ps
    PID TTY          TIME CMD
  46594 pts/5    00:00:00 bash
  71804 pts/5    00:00:00 ps
dennie@HP:~/linux-tut$ 
```

The output shows the current shell process (`bash`) and the `ps` command itself.

To see more detailed information about running processes, you can use the following options:

```python
dennie@HP:~/linux-tut$ ps -ef
UID          PID    PPID  C STIME TTY          TIME CMD
root           1       0  0 07:51 ?        00:00:05 /usr/lib/systemd/systemd --system --deserial
root           2       1  0 07:51 hvc0     00:00:00 /init
root           7       2  0 07:51 hvc0     00:00:00 plan9 --control-socket 7 --log-level 4 --ser
root          44       1  0 07:51 ?        00:00:03 /usr/lib/systemd/systemd-journald
root          92       1  0 07:51 ?        00:00:01 /usr/lib/systemd/systemd-udevd
systemd+     149       1  0 07:51 ?        00:00:00 /usr/lib/systemd/systemd-resolved
systemd+     150       1  0 07:51 ?        00:00:01 /usr/lib/systemd/systemd-timesyncd
root         156
...
```

The `-i` option in the `grep` commands makes the search case-insensitive, so it will match both uppercase and lowercase occurrences of "dennie" and "bash".
