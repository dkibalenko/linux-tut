### Understand the Purpose and Syntax of the `stat` Command
 The `stat` command is used to display detailed information about a file or directory, including its metadata such as permissions, ownership, timestamps, and more.

```python
stat [options] <file_or_directory>
```

The most commonly used options for the `stat` command are:

- `-c` or `--format=<format>`: Specify the output format using a custom format string.
- `-L` or `--dereference`: Follow symbolic links and display information about the target file or directory.
- `-f` or `--file-system`: Display information about the file system instead of the file itself.

```python
stat ~/project/example.txt
```

```python
  File: '/home/labex/project/example.txt'
  Size: 0         	Blocks: 0          IO Block: 4096   regular empty file
Device: 801h/2049d	Inode: 131074      Links: 1
Access: (0644/-rw-r--r--)  Uid: (1000/labex)   Gid: (1000/labex)
Access: 2023-04-11 12:34:56.789012345 +0000
Modify: 2023-04-11 12:34:56.789012345 +0000
Change: 2023-04-11 12:34:56.789012345 +0000
 Birth: -
```

This output provides detailed information about the file, including its size, permissions, ownership, and timestamps.

This is typical output from:

```python
stat /home/labex/project/example.txt
```

It describes the file’s metadata:

```python
File: '/home/labex/project/example.txt'
```

- **File**: The file path being inspected.
- The quotation marks are part of `stat`’s display and help make the path clear.

```python
Size: 0          Blocks: 0          IO Block: 4096   regular empty file
```

- **Size: 0**: The file contains zero bytes, so it is empty.
- **Blocks: 0**: The file currently uses zero filesystem data blocks. An empty file usually does not need a data block.
- **IO Block: 4096**: The filesystem’s preferred block size for input/output is 4096 bytes.
- **regular empty file**:
    - **regular file** means it is an ordinary file, not a directory, symbolic link, device, or socket.
    - **empty** confirms that its size is zero.

```python
Device: 801h/2049d    Inode: 131074      Links: 1
```

- **Device: 801h/2049d**: Identifies the filesystem device where the file is stored.
    - `801h` is the device number in hexadecimal.
    - `2049d` is the same device number in decimal.
- **Inode: 131074**: The file’s unique inode number on that filesystem. The inode stores metadata such as permissions, ownership, timestamps, and pointers to file data.
- **Links: 1**: There is one hard link to this inode—the filename shown in the `File` field. If another **hard link** pointed to the same inode, this number would be greater than 1.

```python
Access: (0644/-rw-r--r--)  Uid: (1000/labex)   Gid: (1000/labex)
```

- **Access: (0644/-rw-r--r--)**: File permissions.
    - `0644` is the octal representation:
        - `0`: No special permission bits.
        - `6`: Owner permissions: read (`4`) + write (`2`) = `6`.
        - `4`: Group permissions: read-only.
        - `4`: Other users’ permissions: read-only.
    - `-rw-r--r--` is the symbolic representation:
        - First `-`: Regular file.
        - `rw-`: Owner can read and write.
        - `r--`: Group can read only.
        - `r--`: Other users can read only.
- **Uid: (1000/labex)**:
    - `Uid` means user ID.
    - User ID `1000` belongs to the user `labex`, who owns the file.
- **Gid: (1000/labex)**:
    - `Gid` means group ID.
    - Group ID `1000` is named `labex`, so the file belongs to the `labex` group.

```python
Access: 2023-04-11 12:34:56.789012345 +0000
```

- **Access** is the last-access time, also called `atime`.
- It records when the file was last read or accessed.
- `2023-04-11`: Date in `YYYY-MM-DD` format.
- `12:34:56`: Time in hours, minutes, and seconds.
- `.789012345`: Nanoseconds.
- `+0000`: Time zone offset from UTC; this is UTC itself.

```python
Modify: 2023-04-11 12:34:56.789012345 +0000
```

- **Modify** is the last-content-modification time, also called `mtime`.
- It changes when the file’s contents are changed.
- Since the file is empty, this timestamp changes if content is added, removed, or replaced.

```python
Change: 2023-04-11 12:34:56.789012345 +0000
```

- **Change** is the last-metadata-change time, also called `ctime`.
- It changes when file metadata changes, such as:
    - Permissions
    - Owner or group
    - Hard links
    - File contents
- On Linux, this is **not** the file creation time.

```python
Birth: -
```

- **Birth** represents the file creation time, also called `btime`.
- `-` means the filesystem or operating system does not provide a creation-time value for this file.

A useful distinction is:

|Field|Meaning|
|---|---|
|`Access`|Last time the file was read|
|`Modify`|Last time the file contents changed|
|`Change`|Last time the file’s metadata changed|
|`Birth`|File creation time, if supported|

For example:

```python
chmod 600 example.txt
```

would change `Change` because the permissions changed, but it would not change `Modify` because the file contents stayed the same.

### Retrieve File Metadata Using the `stat` Command
Use the `-c` or `--format` option to customize the output format of the `stat` command.

```python
dennie@HP:~/linux-tut$ stat -c '%s %y' labex/sample_file.txt 
```

```python
0 2026-09-11 20:14:23.557706547 +0000
```

This allows you to extract specific metadata fields that are most relevant to your needs.

#### Analyze File Permissions and Ownership with `stat`

```python
dennie@HP:~/linux-tut$ stat -c '%A %u %G' labex/project/numbers.txt 
-rw-r--r-- 1000 dennie
```

- The file permissions: `-rw-r--r--` (read-write for the owner, read-only for the group and others)
- The user ID (UID) of the file owner: `1000` (which corresponds to the `labex` user)
- The owner file: `dennie`

Long-form options to get the same information:

```python
dennie@HP:~/linux-tut$ stat --format='%A %U %G' labex/project/numbers.txt 
-rw-r--r-- dennie dennie
```

This shows the same information, but with the **user** and **group names** instead of the numeric IDs.

### Output Customization

The `stat` command can customize output with `-c` (or `--format`). 
The format is a quoted string containing placeholders that begin with `%`.

Assuming:

```python
dennie@HP:~/linux-tut$ FILE=labex/sample_file.txt
```

**Common examples**

Show only the file size:

```python
dennie@HP:~/linux-tut$ stat -c '%s' "$FILE"
0
```

Show size with a label:

```python
dennie@HP:~/linux-tut$ stat -c 'Size: %s bytes' "$FILE"
Size: 0 bytes
```

Show permissions:

```python
dennie@HP:~/linux-tut$ stat -c '%A' "$FILE"
-rw-r--r--
```

Show the numeric form:

```python
dennie@HP:~/linux-tut$ stat -c '%a' "$FILE"
644
```

Show owner and group:

```python
dennie@HP:~/linux-tut$ stat -c 'Owner: %U, Group: %G' "$FILE"
Owner: dennie, Group: dennie
```

Show numeric user and group IDs:

```python
dennie@HP:~/linux-tut$ stat -c 'UID: %u, GID: %g' "$FILE"
UID: 1000, GID: 1000
```

Show inode and link count:

```python
dennie@HP:~/linux-tut$ stat -c 'Inode: %i, Links: %h' "$FILE"
Inode: 59548, Links: 1
```

Show file type:

```python
dennie@HP:~/linux-tut$ stat -c 'Type: %F' "$FILE"
Type: regular empty file
```

Show access, modification, and change times:

```python
dennie@HP:~/linux-tut$ stat -c 'Access: %x' "$FILE"
Access: 2026-09-11 20:14:23.557706547 +0000
```

```python
dennie@HP:~/linux-tut$ stat -c 'Modified: %y' "$FILE"
Modified: 2026-09-11 20:14:23.557706547 +0000
```

```python
dennie@HP:~/linux-tut$ stat -c 'Changed: %z' "$FILE"
Changed: 2026-09-11 20:14:23.557706547 +0000
```

The shorter timestamp forms are:

```python
dennie@HP:~/linux-tut$ stat -c 'Access: %X' "$FILE"    # Unix timestamp
Access: 1789157663
```

```python
dennie@HP:~/linux-tut$ stat -c 'Modified: %Y' "$FILE"  # Unix timestamp
Modified: 1789157663
```

```python
dennie@HP:~/linux-tut$ stat -c 'Changed: %Z' "$FILE"   # Unix timestamp
Changed: 1789157663
```

Display several fields in one line:

```python
dennie@HP:~/linux-tut$ stat -c 'Name: %n | Size: %s bytes | Permissions: %A | Owner: %U:%G' "$FILE"
Name: labex/sample_file.txt | Size: 0 bytes | Permissions: -rw-r--r-- | Owner: dennie:dennie
```

Create a CSV-style output:

```python
dennie@HP:~/linux-tut$ stat -c '"%n",%s,%a,%U,%G,%y' "$FILE"
"labex/sample_file.txt",0,644,dennie,dennie,2026-09-11 20:14:23.557706547 +0000
```

This is useful for scripts or importing data into spreadsheets.

#### Create a tab-separated output

```python
dennie@HP:~/linux-tut$ stat -c '%n\t%s\t%A\t%U\t%G' "$FILE"
labex/sample_file.txt\t0\t-rw-r--r--\tdennie\tdennie
```

Include filesystem block information:

```python
dennie@HP:~/linux-tut$ stat -c 'Size: %s | Blocks: %b | Block size: %B' "$FILE"
Size: 0 | Blocks: 0 | Block size: 512
```

- `%b`: Number of blocks allocated
- `%B`: Size of each allocated block

#### Inspect multiple files

```python
dennie@HP:~/linux-tut$ stat -c '%n: %s bytes, modified %y' labex/project/*.txt
```

```python
labex/project/numbers.txt: 197 bytes, modified 2026-09-05 18:44:26.448570173 +0000
labex/project/task2_output.txt: 50 bytes, modified 2026-09-05 17:21:37.686189341 +0000
labex/project/task4_output.txt: 40 bytes, modified 2026-09-05 18:55:05.230046631 +0000
```

The `%n` placeholder prints the filename, which is important when processing multiple files.

#### Useful format placeholders

|Placeholder|Meaning|
|---|---|
|`%n`|File name|
|`%N`|Quoted file name, including symlink information|
|`%s`|File size in bytes|
|`%b`|Number of allocated blocks|
|`%B`|Block size|
|`%a`|Permissions in octal, such as `644`|
|`%A`|Permissions in symbolic form, such as `-rw-r--r--`|
|`%F`|File type|
|`%i`|Inode number|
|`%h`|Hard-link count|
|`%u`|Numeric user ID|
|`%U`|User name|
|`%g`|Numeric group ID|
|`%G`|Group name|
|`%x`|Human-readable access time|
|`%y`|Human-readable modification time|
|`%z`|Human-readable metadata-change time|
|`%X`|Access time as Unix timestamp|
|`%Y`|Modification time as Unix timestamp|
|`%Z`|Metadata-change time as Unix timestamp|

You can also use `--printf` instead of `-c`:

```python
dennie@HP:~/linux-tut$ stat --printf='Name: %n\nSize: %s bytes\nOwner:
```

```python
%U\n' "$FILE"
Name: labex/sample_file.txt
Size: 0 bytes
Owner: dennie
```

With `--printf`, use `\n` for a newline and `\t` for a tab.
