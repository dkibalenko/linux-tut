## MAC Address
The MAC (Media Access Control) is an identifier that every network device uses to **uniquely identify itself** on a network.

The MAC adress is also referred to as the **physical** or **hardware address**.

No two devices in the world may have the same MAC address.

It's made up of 6 bytes hexadecimal number that is **burned into every NIC** (network interface card).

The MAC address can contain any number and it also contains alphabets from A to F.

The MAC address is broken up into two parts:
- the first 3 bytes identifies the **manufacturer of the NIC (network interface card)**, such as TPlink, Lingsys, etc.
- the last 3 bytes are a **unique number** from the manufacturer, that **identifies each device on a network**.

#### MAC Formats

**Linux/Apple**

---

### MAC Address Purpose
The purpose of the MAC address is so network devices can **communicate with each other**. This communication is ultimately done using the MAC address and this is whether the **devices are close to each other** such as being on the **same network**, or if they are thousands of miles apart on a different network.

> The bottom line is that devices **communicate with each other** using the **MAC address**.

---

### Difference from IP Address
Why using IP address though?

Public IP addresses are **unique** but public and private IP addresses can **periodically change**.

You may have had a certain IP address for the past few months but an Internet service provider or a network administrator can change your IP to a different one, periodically. MAC addresses **don't change** they are **permanent**.

---

### TCP/IP work with IP Address & MAC Address
The way TCP/IP works which is the language that is used on networks the Internet, a networking device needs **both an IP address and a MAC address**.

The IP address and MAC address **work together** so devices can talk to each other.

The IP is used to **locate a device**.

For example, a person lives in a house and has a **name** and a **mailing address** of the house. The mailing address tells us the country, city, and the street of **where the house is located**. But it doesn't necessarily tell us who lives in the house.

But the MAC address is like the **name of the person**. It tell us **who lives in the house**.

### LAN Communication
#### Scenario 1 (LAN)
The computer A wanted to communicate with computer B on this local network.

1. Computer A is going to inspect computer B's IP address to see if it's on the same network or not.

2. After inspection computer A now knows that computer B is on the same network because the **IP addresses are in the same group**.

3. In order for communication to take place computer A needs computer B's MAC address. And it finds this by sending out what's called an **ARP (Address Resolution Protocol) broadcast**.

4. It will broadcast to **every device** on its local network, asking computer B to **identify itself** by asking for its MAC address.

5. Once computer B tells Computer A its MAC address communication can finally take place.

#### Scenario 2 (Different Nerworks)
Computer A wants to go to `google.com`.

1. Computer A needs the MAC address for Google's web server. The problem is it doesn't know what it is. This is why **it needs the IP address to get the web servers MAC address**.

2. Computer A will type in `google.com` in its web browser and then DNS will change `google.com` into an IP address, so computers can understand it. Computer A will inspect the IP address and realize that the **IP is not in its local network** because the IP address is not in the same group as the computer its local network. Computer A realizes that Google's IP address is **on a different network**.

3. Since it's on a different network it's going to **forward that data to its default gateway**, which is thr router. It will let the router deal with.

4. Computer A will send out an **ARP broadcast**. This time it will ask for the MAC address of the **defaul gateway**. Once it has the MAC address it will send the data to the default gateway.

5. Once the default gateway has the data it will inspect Google's IP address and **determine the best PATH** for the data to reach the destination. It will forward the data to the **next router**.But before it can forward the data it also needs the **next router's MAC address**. It also going to do an ARP broadcast and ask for that router's MAC address. Once it has it will forward the data, and so on.

6. Once the final router in its path That rotor needs the MAC address of the Google's web server. Once it the data has reached its final destination.

> Each time the data is passed between a computer or a router it **uses the MAC address** to forward the data to each step.

---

### Summary

1. The IP address is used to **locate** and **get to** the final destination.
2. The MAC address is **used at each step on its way** to the final destination.
