## IP Address
[Claudflare Article](https://www.cloudflare.com/learning/dns/glossary/what-is-my-ip-address/)

IP addresses are unique identifiers assigned to a **device** or **domain** used to determine who is who on the Internet. IP addresses can be formatted differently depending on whether they use IPv4 or IPv6 protocol.

Via **DNS resolvers**, which translate human-readable domain names into IP addresses, users are able to access websites without memorizing this complex series of characters. Each IP packet will contain both the IP address of the device or domain sending the packet and the IP address of the intended recipient, much like how both the destination address and the return address are included on a piece of mail.

- An IP address is a numeric address
- It's an identifier for a computer or device on a network

Every device has to have an IP address for communication purposes.

### Two Types of IPv4 addresses

#### Public - Unique
When you order an Internet service from ISP to your home they are going to assign to your **router** a **Public IP** address. This public IP address is **registered on the Internet**. It's what gives you access to the WWW. If you don't have the Public IP you **don't have access to the Internet**.

Public IPs are **unique**. There are **no duplicates** in the world of public IPs.

- Unique
- Publicly registered on the Internet
- Used externally
- Assigned by an ISP
- Not free
- Not secure (Use a Proxy or VPN, which hides your public IP)

#### Private - Non-Unique
In order to prevent a shortage of public IPs it was developed **Private IPs**.

Private IPs are **not publicly registered** on the Internet. You **can't access** the Internet using a private IP. However, if you have a private IP and you want to acces the Internet, your private IP **must be converted** into a public IP **before** you can access the Internet.
> This is because of Private IPs are **only used internally** such as inside a home or business. They are **not used out** on the public Internet.

- Non-Unique. Can be used on other private networks
- Not publicly registered
- Used internally
- Assigned by a router
- Free
- More secure, since they are hidden from the Internet

#### DHCP service
The service inside your router called DHCP is what **assigns** your internal devices a **private IPs**.

In order to **not run out of public IPs** we have routers that assign the devices **inside** our home or business **private IPs**.

#### NAT
When these devices need to access the Internet their private IP addresses will be **translated to** the **one public IP address** that we have been given.
The service that translates this is called **NAT** or **Network Address Translation** and it's a service that's **built into** our router.

Not only does it translate **private to public** but it also translates **public to private**.

If a computer on the Internet wants to communicate with the computer **on this private network**, the public IP address needs to be translated by NAT **to the private IP address** for that computer.

#### Private IP Classes
![Private IP Classes](../images/private_ip_address_ranges.png)
![Subnet Masks & IP Classes](../images/IP-Subnet-Masks.png)

1. **Class A**. This class is typically used for **large** organizations:
    - Networks $2^8 = 256$; Hosts $2^24 = 16,777,216$
2. **Class B**. This class is typically used in **medium-sized** orgs.
3. **Class C**. This is used in **small** orgs or homes.

Class C is the **most popular** range of private IP addresses, starting with `192.168.`.

### Dynamic vs Static vs Private/Public IPs
**Public/private** describes **where** the address is valid.

**Dynamic/static** describes **how** it was assigned.

Any combination is possible.


||Dynamic (DHCP hands it out, can change)|Static (fixed, set manually or reserved)|
|---|---|---|
|**Private**|Your `192.168.1.105`, assigned by the router's DHCP|A printer or NAS set to `192.168.1.50` permanently|
|**Public**|Your `37.188.205.146`, probably, since home ISPs usually rotate it|A server's address, paid for as a "static IP" from the ISP|

## How to tell which one you have

- **Private IP, dynamic or static?** Check the DHCP flag:
    
```powershell
PS C:\Users\denni> Get-NetIPInterface -InterfaceAlias "Wi-Fi" -AddressFamily IPv4 | Select InterfaceAlias, Dhcp

InterfaceAlias    Dhcp
--------------    ----
Wi-Fi          Enabled
```
    
`Enabled` means dynamic. `Disabled` means static.
- **Public IP, dynamic or static?** You can't tell from the PC. Check whether it changes after the router reboots, or ask the ISP. Home plans are usually dynamic.

## Why support cares

- **Dynamic private IP:** a device's address can change after a reboot or lease expiry. Symptom: "the printer worked yesterday, today it doesn't". Fix: a DHCP reservation on the router.
- **Dynamic public IP:** allowlists and remote access break when the ISP rotates the address. Symptom: "I'm blocked, but I was allowed last week". Fix: a static IP from the ISP, or DDNS.

**WHAT TO LOOK FOR:** Servers, printers, and anything others connect _to_ should be static or reserved. Laptops and phones are fine as dynamic.
