## PILLAR 1 (backfill): Filesystem Navigation & Discovery

**Support context:** Half of support work is "where is the thing, and how did it get there." These commands are how you _find_ files/directories fast, without guessing paths — critical when you've never seen a customer's box before.

---

### Absolute vs. Relative Paths

You already navigate fine, but the distinction matters _specifically_ in support because scripts and configs break when this is confused.

- **Absolute path**: starts with `/` — unambiguous, always means the same thing regardless of where you're standing. `/var/log/syslog`.
- **Relative path**: starts from wherever you currently are (`pwd`). `../log/syslog`, `./config.yaml`.

**Developer bridge:** exactly like Python's `os.path` absolute vs. relative imports, or Django's `BASE_DIR / "static"` pattern — a relative path is only meaningful in context, an absolute one isn't.

**GOTCHA:** cron jobs and systemd services often fail silently because they run with a _different_ working directory than your interactive shell — a script that works fine when you run it manually (using relative paths) can fail when cron runs it (different `pwd`). This is one of the single most common "why does it work manually but not automated" tickets.
> **Support rule: scripts meant to run unattended should use absolute paths, always.**

### `find` — the most important discovery tool in this pillar

**PURPOSE (support):** "Where is this file" and "which files match this pattern" — by name, size, age, type, or permission — across an entire directory tree.

**SYNTAX:** `find [path] [conditions] [action]`

|Flag|Finds by|
|---|---|
|`-name "pattern"`|filename (case-sensitive), supports wildcards|
|`-iname "pattern"`|filename, case-**in**sensitive|
|`-type f` / `-type d` / `-type l`|file / directory / symlink|
|`-size +100M`|files larger than 100MB|
|`-mtime -1`|modified in the last 1 day|
|`-mmin -30`|modified in the last 30 minutes|
|`-perm 777`|exact permission match|
|`-user dennie`|owned by a specific user|

**READ THE OUTPUT:** the `2>/dev/null` on that last one isn't decoration — searching from `/` will hit directories you don't have permission to read (like `/root`), and `find` will spam `Permission denied` to stderr for each one. ### `locate` — instant search, but only as fresh as its last index

**PURPOSE (support):** same idea as `find -name`, but reads from a **prebuilt database** instead of walking the live filesystem — near-instant results.

````python
dennie@HP:~/linux-tut$ locate sshd_config
/mnt/c/Program Files/Git/etc/ssh/sshd_config
/mnt/c/Windows/servicing/LCU/Package_for_RollupFix~31bf3856ad364e35~amd64~~19041.6466.1.0/amd64_openssh-server-components-onecore_31bf3856ad364e35_10.0.19041.5737_none_9bae0398d266ff1a/f/sshd_config_default
````

**GOTCHA — the one that actually matters:** that database (`updatedb`) typically only refreshes once a day via cron. If a file was created in the last few hours, `locate` **won't find it** — and it won't tell you it's out of date, it'll just come back empty. **Support rule: `locate` for "where does this normally live" on stable files; `find` when you need the ground truth right now** (e.g., a file created 10 minutes ago as part of the incident you're investigating).

### `file` — identify what something actually is
PURPOSE (support): Linux doesn't care about file extensions the way Windows does — a file named `data.txt` could secretly be a JPEG. `file` inspects the actual content (magic bytes) to tell you the real type.

````python
dennie@HP:~/linux-tut$ file /etc/passwd
/etc/passwd: ASCII text

dennie@HP:~/linux-tut$ file /bin/ls
/bin/ls: ELF 64-bit LSB pie executable, x86-64, version 1 (SYSV), dynamically linked, interpreter /lib64/ld-linux-x86-64.so.2, BuildID[sha1]=05dad2c279f7651722809fa75adba6bf9ab1c209, for GNU/Linux 3.2.0, stripped

dennie@HP:~/linux-tut$ file /usr/lib/wsl/drivers
/usr/lib/wsl/drivers: directory
````

**TICKET SCENARIO:** _"User uploaded a file, app rejected it as invalid format."_ `file suspicious_upload` tells you what it actually is, regardless of what extension the user (or an attacker) gave it — extension-based validation alone is exactly how this class of bug/security issue happens.

### TRY IT

````python
dennie@HP:~/linux-tut$ ls -ltr /var/log | tail -5
-rw-r-----  1 syslog    adm             124798 Aug 24 17:11 kern.log
drwxr-xr-x  2 root      root              4096 Aug 24 18:09 apt
-rw-r--r--  1 root      root            492454 Aug 24 18:09 dpkg.log
-rw-r-----  1 syslog    adm              12650 Aug 24 18:09 auth.log
-rw-r-----  1 syslog    adm             374713 Aug 24 18:09 syslog
````

Exactly the pattern intended: `syslog` and `auth.log` at the very bottom, both timestamped `18:09` — the most recently modified. If you were investigating "something broke around 18:09," these are your first two stops.

````python
dennie@HP:~/linux-tut$ find /etc -name "*.conf" | head -5
find: ‘/etc/ssl/private’: Permission denied
/etc/dbus-1/system.d/com.ubuntu.SoftwareProperties.conf
/etc/rsyslog.conf
/etc/fuse.conf
/etc/sysctl.d/10-zeropage.conf
/etc/sysctl.d/10-ptrace.conf
````

This isn't a bug in your command — it's `find` correctly reporting that it can't descend into `/etc/ssl/private` as a regular user. **That's actually the system working correctly**: private TLS keys should never be world-readable, so `find` (running as you) gets locked out, prints a warning to stderr, and keeps going with everything else. This is the `2>/dev/null` pattern from before in action — without it, every locked-out directory prints noise like this alongside your real results.

````python
dennie@HP:~/linux-tut$ file /etc/passwd /bin/ls
/etc/passwd: ASCII text
/bin/ls:     ELF 64-bit LSB pie executable, x86-64, version 1 (SYSV), dynamically linked, interpreter /lib64/ld-linux-x86-64.so.2, BuildID[sha1]=05dad2c279f7651722809fa75adba6bf9ab1c209, for GNU/Linux 3.2.0, stripped
````

Exactly the point — `/etc/passwd` has no extension at all, and `/bin/ls` isn't `ls.exe` or anything extension-based. `file` reads the actual bytes and tells you the truth: one's plain text, one's a compiled binary. This is why "just check the file extension" is not a valid Linux support strategy.
